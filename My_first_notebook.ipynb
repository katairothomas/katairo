## Plotting

```python
x = np.random.binomial(n=30, p=0.5, size=10_000) ##Simulate flipping a fair coin 30 times, repeat that experiment 10,000 times, and record how many heads you get each time
x
```

```python
import plotly.express as px
```

```python
fig1 = px.histogram(
    x=x,
    title= " Histogram of randomly generated binomial data",
    labels={'x': "Number of Heads", 'y': "Frequency"},
    nbins=30)
fig1.show()
```
