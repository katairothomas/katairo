## We are now going to use pivot tables

```python
pivot_country_year_taxon = (
    df_samples
    .pivot_table(
        index= ["country", "year"],
        columns="taxon",
        values="sample_id",
        aggfunc="count",
        fill_value=0
    )
    )
pivot_country_year_taxon
```

```python
pivot_location_year_taxon_bf = (
    df_samples
    .query("country == 'Burkina Faso'")
    .pivot_table(
        index= ["admin1_name", "admin2_name", "location", "year"],
        columns="taxon",
        values="sample_id",
        aggfunc="count",
        fill_value=0
    )
)
pivot_location_year_taxon_bf
```

# Plotting maps of sampling locations

```python
!pip install -q ipyleaflet
```

```python
from ipyleaflet import Map, Marker, basemaps, ScaleControl
```

```python
m = Map(
    basemap=basemaps.OpenStreetMap.Mapnik,
    center=[0, 20],
    zoom=3,
)
m
```

```python
pivot_location_taxon = (
    df_samples
    .pivot_table(
        index= ["country", "location", "latitude", "longitude"],
        columns="taxon",
        values="sample_id",
        aggfunc="count",
        fill_value=0
    )
)
pivot_location_taxon
```

```python
m = Map(
    basemap=basemaps.OpenStreetMap.Mapnik,
    center=[0, 20],
    zoom=3,
)

#Use for loop to add sampling locations
for row in pivot_location_taxon.reset_index().itertuples():
  title = (
      f"{row.location}, {row.country} ({row.latitude:3f}.{row.longitude:3f})\n"
      f"{row.gambiae} gambiae, {row.coluzzii} coluzzi, ({row.arabiensis} arabiensis"
  )
  marker = Marker(
      location=(row.latitude, row.longitude),
      draggable=False,
      title=title,
  )
  m.add_layer(marker)

##Add scale bar
m.add_control(ScaleControl(position="bottomleft"))

##Display map
m
```

### Number of samples available in the Ag3.0 resource that were collected in Mali, summarized by year, location and taxon

```python
samples_Mali_year_location = \
df_samples \
.query("country == 'Mali'") \
.pivot_table(
    index=["year", "location", "country"],
    columns="taxon",
    values="sample_id",
    aggfunc="count"
)

samples_Mali_year_location
```

### How many countries are there for which we have samples of A. coluzzii

```python
countries_coluzzii = (
    df_samples
    .query("taxon == 'coluzzii'")
    .pivot_table(
        index="country",
        values="sample_id",
        aggfunc="count"
    )
)
countries_coluzzii
```
